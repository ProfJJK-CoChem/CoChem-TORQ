## Working with your molecule
Paste one XYZ geometry (coordinates in Å), then explicitly set charge, multiplicity, recipe and requested products. You can also paste a complete TORQ JSON request from BASE or import a prepared TOPOS handoff with the application.
**Import and preview** labels the geometry as supplied input. Rotate and tilt the actual mass-centered principal-axis view. **Check request** checks state, method, product and resource compatibility before **Submit to Actions**.
The HF teaching recipe illustrates the workflow; its small basis is unsuitable for high-accuracy identification. Exact revDSD and later spectroscopy stages remain unavailable until their independent qualifications pass.

## Recovering results
After submission, save the run ID. **Refresh status** reads the actual Actions state. **Download results** retrieves the run artifacts into the selected directory.
Inspect the result manifest, scientific stage statuses, provenance and uncertainty. A completed Actions job may contain a partial calculation. Missing harmonic or anharmonic data do not imply zero corrections, and equilibrium constants are not vibration-corrected ground-state constants.
You can close Codespaces while Actions runs and enter the saved run ID in a later session.
