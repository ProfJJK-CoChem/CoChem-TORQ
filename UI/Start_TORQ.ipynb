## Phase 3 (Stage 2.0 - 2.1): Conformer & Binding-Site Isomer Search (GOAT / CREST Union)

```python
from Libraries.cochem_torq_mace import evaluate_physical_potential
from Libraries.cochem_torq_goat import ConformerRecord, deduplicate_stage_b_spectroscopic

# Evaluate Machine Learning Fast Filtering (MLFF) Potential
mlff_energy_ev, mlff_forces, mlff_converged = evaluate_physical_potential(
    symbols=validated_symbols,
    coordinates=aligned_coords
)
mlff_energy_hartree = float(mlff_energy_ev / 27.211386245988)

# Construct initial conformer record
initial_conformer = ConformerRecord(
    index=0,
    symbols=validated_symbols,
    coordinates=aligned_coords.tolist(),
    energy_hartree=mlff_energy_hartree,
    energy_kcal_rel=0.0,
    rotational_constants_mhz=(A_mhz, B_mhz, C_mhz),
    rotational_constants_ghz=(A_ghz, B_ghz, C_ghz),
    inertial_defect_u_a2=delta_defect,
    planar_moments_u_a2=planar_moments,
    ray_asymmetry_kappa=ray_kappa,
    origin_engine="TORQ-AUTOCRATIC-UI",
    seed_id="seed_00"
)

# Deduplicate conformers via Stage B Spectroscopic Windowing
conformer_survivors = deduplicate_stage_b_spectroscopic([initial_conformer])

display(HTML(f"""
<div style="background-color: #0f172a; border-left: 4px solid #f59e0b; padding: 12px; border-radius: 4px; font-family: monospace; color: #f8fafc;">
    <span style="color: #f59e0b; font-weight: bold;">[CONFORMER ENUMERATION & SPECTROSCOPIC FILTERING COMPLETE]</span><br>
    <b>MLFF Potential Energy:</b> {mlff_energy_hartree:.8f} E_h ({mlff_energy_ev:.4f} eV)<br>
    <b>MLFF Gradient Convergence:</b> <span style="color: {'#22c55e' if mlff_converged else '#f59e0b'}; font-weight: bold;">{'CONVERGED' if mlff_converged else 'PRE-RELAXED'}</span><br>
    <b>Unique Spectroscopic Conformers Retained:</b> {len(conformer_survivors)}<br>
    <b>Deduplication Metric:</b> ΔB/B = 2ΔR/R (<0.1% rotational variance window)
</div>
"""))
```

## Phase 4/5 (Stage 3.0 - 4.5): Method Matrix Cascade Optimization & Frozen-Monomer Protections

- **Hessian Preconditioning:** `InHess XTB2` / `Lindh` (`Calc_Hess true` strictly prohibited).
- **Dispersion Mandate:** Mandatory D3/D4 dispersion correction for all DFT weak-complex optimizations.

```python
from Libraries.torq_config import TorqRunParams
from Libraries.cochem_torq_engine import ExecutionContext, route_cascade_rules

# Tier selection and quantum chemistry parameters
torq_config = TorqRunParams(
    tier="T3-3h",
    wall_time_tier="T3-3h",
    engine="ORCA",
    method="wB97M-V",
    basis_set="def2-TZVPP",
    dispersion="D4",
    bsse_correction="counterpoise",
    keywords=["RIJCOSX", "TightOpt", "TightSCF", "DEFGRID3", "InHess XTB2"]
)

exec_context = ExecutionContext()
dispatch_payload = route_cascade_rules(
    point_coords=aligned_coords,
    context=exec_context,
    symbols=validated_symbols,
    charge=active_payload.get("charge", 0),
    multiplicity=active_payload.get("multiplicity", 1),
    method=torq_config.method,
    basis_set=torq_config.basis_set,
    is_complex=is_complex,
    counterpoise=True,
    extra_options="! TightOpt TightSCF DEFGRID3 InHess XTB2"
)

display(HTML(f"""
<div style="background-color: #0f172a; border-left: 4px solid #ec4899; padding: 12px; border-radius: 4px; font-family: monospace; color: #f8fafc;">
    <span style="color: #ec4899; font-weight: bold;">[METHOD MATRIX CASCADE CONFIGURATION]</span><br>
    <b>Recommended Tier:</b> <code>{torq_config.tier} (Recipe R2: Frozen Monomers + ωB97M-V/TZ + CP + VPT2)</code><br>
    <b>Electronic Structure Level:</b> <code>{torq_config.method} / {torq_config.basis_set}</code><br>
    <b>Dispersion Correction:</b> <span style="color: #22c55e; font-weight: bold;">D4 (Mandatory for Weak Complexes)</span><br>
    <b>BSSE Treatment:</b> <span style="color: #22c55e; font-weight: bold;">Boys-Bernardi Counterpoise (CP)</span><br>
    <b>Grid Optimization Strategy:</b> Dynamic Cascade (defgrid1 -> defgrid3)<br>
    <b>Hessian Preconditioner:</b> InHess XTB2 (Calc_Hess true forbidden)<br>
    <b>Intermolecular TolMaxG:</b> 1.00e-05 E_h/a_0
</div>
"""))
```
